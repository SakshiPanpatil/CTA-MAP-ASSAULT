```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("Assualt data.csv")

# Preview the data
print("Shape:", df.shape)
df.head()
```

```python
# Convert text columns to lowercase for consistency
df = df.applymap(lambda x: x.lower() if isinstance(x, str) else x)

# Filter rows that contain 'cta'
cta_df = df[df.apply(lambda row: row.astype(str).str.contains('cta', case=False, na=False).any(), axis=1)]

print("CTA-only data shape:", cta_df.shape)
cta_df.head()
```

```python
# Further filter only rows that mention 'bus'
cta_bus_df = cta_df[cta_df.apply(lambda row: row.astype(str).str.contains('bus', case=False, na=False).any(), axis=1)]

print("CTA + Bus data shape:", cta_bus_df.shape)
cta_bus_df.head()
```

```python
cta_bus_df = cta_bus_df[
    (cta_bus_df['Agency'].str.lower() == 'chicago transit authority') &
    (cta_bus_df['Mode Name'].fillna(df['Mode']).str.lower() == 'bus')
].copy()
```

```python
# Drop duplicates
cta_bus_df = cta_bus_df.drop_duplicates()

# Optionally reset index
cta_bus_df = cta_bus_df.reset_index(drop=True)

# Save cleaned data
cta_bus_df.to_csv("Cleaned_CTA_Bus_Data.csv", index=False)

print("Cleaned CTA + Bus dataset saved as 'Cleaned_CTA_Bus_Data.csv'")
cta_bus_df.info()
```

```python
import pandas as pd
from pathlib import Path

source = Path("notebook\Assualt data.csv")
df = pd.read_csv(source, low_memory=False)

agency = df["Agency"].astype(str).str.strip().str.lower()
mode = (
    df.get("Mode Name", df["Mode"])
      .fillna(df["Mode"])
      .astype(str)
      .str.strip()
      .str.lower()
)
cta_bus_df = (
    df[(agency == "chicago transit authority") & (mode == "bus")]
    .drop_duplicates()
    .reset_index(drop=True)
)

cta_bus_df.to_csv("Cleaned_CTA_Bus_Data.csv", index=False)
print("Saved CTA bus-only dataset with", len(cta_bus_df), "rows")
```
